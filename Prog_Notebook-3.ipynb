## <span style="color:darkgreen"> 1) Operações Aritiméticas Elementares </span>

Python funciona de modo análogo a uma calculadora. Os operadores fundamentais de adição, subtração e multiplicação seguem a notação algébrica convencional:

a) Adição (+)

b) Subtração (-)

c) Multiplicação (*)


A criação de um código com essas operações é bastante simples, conforme mostra o exemplo 1.1.

```python
# Exemplo 1.1

# 1. Adição (+)
# Exemplo: Agregação de Custo Fixo (1500) e Custo Variável (3200)
print(1500 + 3200)

# 2. Subtração (-)
# Exemplo: Lucro Econômico = Receita Total (8000) menos Custo Total (4700)
print(8000 - 4700)

# 3. Multiplicação (*)
# Exemplo: Receita Total = Preço Unitário (45.50) multiplicado pela Quantidade (120)
print(45.50 * 120)
```

## <span style="color:darkgreen"> 2) Divisão: Decimal, Inteira e Resto </span>

A linguagem Python possui operadores específicos para tratar o quociente e o resto de uma operação de divisão:

a) Divisão convencional (/): retorna sempre um número decimal (float), independentemente de a divisão ser exata ou não.


b) Divisão inteira (//): descarta a parte fracionária e retorna apenas o quociente inteiro da operação. É aplicada na quantificação de bens indivisíveis ou na definição de lotes de transporte.


c) Resto da divisão ou módulo (%): retorna o resíduo resultante de uma divisão inteira. Útil para verificar paridade, periodicidades cíclicas ou sobras de estoques.

Vejamos como esses operadores funcionam na prática por meio do Exemplo 2.1.

```python
# Exemplo 2.1

# 1. Divisão convencional de ponto flutuante (/)
# Exemplo: Custo Médio = Custo Total (1000) dividido pela Produção (8 unidades)
print(1000 / 8)

# 2. Divisão inteira (//)
# Exemplo: Capacidade de 250 toneladas transportadas em carretas de 30 toneladas
print(250 // 30)

# 3. Resto da divisão ou módulo (%)
# Exemplo: Carga restante que não preenche uma carreta inteira
print(250 % 30)
```

## <span style="color:darkgreen"> 3)  Exponenciação (ou Potenciação) </span>

A exponenciação é representada em Python pelo operador de duplo asterisco (**). Esta operação é essencial para cálculos de matemática financeira, juros compostos e funções de produção do tipo Cobb-Douglas. O Exemplo 3.1 nos mostra como codificar essa operação.

```python
# Exemplo 3.1

# Potenciação (**)
# Exemplo: Fator de capitalização de juros compostos: (1 + i)^t
# Taxa de 5% ao mês durante 12 meses: (1.05)**12
print(1.05 ** 12)

# Exemplo: Função de Produção do tipo Y = K^0.4
# Para um estoque de capital K = 100
print(100 ** 0.4)
```

## <span style="color:darkgreen"> 4)  Radiciação </span>

A raiz de um número pode ser calculada de duas formas em Python:

1) Por propriedade algébrica da potenciação: sabendo que $\sqrt[n]{x} = x^{1/n}$, utiliza-se o operador ** com expoente fracionário. Deve-se isolar a fração entre parênteses para assegurar a ordem correta de precedência.

2) Por meio da biblioteca math: para a raiz quadrada, utiliza-se a função especializada math.sqrt().

O Exemplo 4.1 mostra essas opções.

```python
# Exemplo 4.1

import math

# 1. Radiciação via operador de potência (**)
# Raiz quadrada de 144: 144^(1/2)
print(144 ** (1 / 2))

# Raiz cúbica de 1000: 1000^(1/3)
print(1000 ** (1 / 3))

# 2. Radiciação via função math.sqrt()
# Exemplo: Cálculo do desvio-padrão a partir de uma variância igual a 64
print(math.sqrt(64))
```

## <span style="color:darkgreen"> 5)  Logaritmos </span>

O logaritmo é empregado na Economia para linearizar modelos, calcular taxas contínuas de retorno, estimar elasticidades e ajustar séries temporais de alta volatilidade.

Para operações logarítmicas, utiliza-se a biblioteca padrão math:

a) math.log(x): retorna o logaritmo natural de $x$ (base $e$, onde $e \approx 2,71828$).

b) math.log10(x): retorna o logaritmo de $x$ na base 10.

c) math.log(x, base): retorna o logaritmo de $x$ em uma base arbitrária especificada no segundo argumento.

Vejamos como essa biblioteca funciona por meio do Exemplo 5.1.

```python
# Exemplo 5.1

import math

# 1. Logaritmo natural (base e)
# Exemplo: Utilizado no cálculo de retornos contínuos de ativos
print(math.log(100))

# 2. Logaritmo na base 10
print(math.log10(1000))

# 3. Logaritmo em base customizada (por exemplo, base 2)
print(math.log(32, 2))
```

## <span style="color:darkgreen"> 6)  Utilização de Variáveis para Cálculos Econômicos </span>

Atribuir valores a variáveis permite organizar, reutilizar e documentar a lógica do cálculo. O operador de atribuição em Python é o sinal de igualdade (=).

Uma variável deve possuir nomes claros e representativos do seu significado econômico, respeitando as boas práticas de programação:

1) Nomes em letras minúsculas, separados por sublinhado (snake_case).

2) Ausência de caracteres especiais ou acentuação no identificador.

```python
import math

# Exemplo 1: Receita, Custo e Lucro Operacional

preco_unitario = 120.0
quantidade_vendida = 450
custo_fixo = 12000.0
custo_variavel_unitario = 75.0

receita_total = preco_unitario * quantidade_vendida
custo_total = custo_fixo + (custo_variavel_unitario * quantidade_vendida)
lucro_operacional = receita_total - custo_total

print("--- Exemplo 1: Resultados Financeiros ---")
print("Receita Total: R$", receita_total)
print("Custo Total: R$", custo_total)
print("Lucro Operacional: R$", lucro_operacional)
```

```python

# Exemplo 2: Matemática Financeira (Juros Compostos)
# Formula: M = C * (1 + i)^t

capital_inicial = 50000.0
taxa_juros_anual = 0.085  # 8,5% ao ano
periodo_anos = 5

montante_acumulado = capital_inicial * ((1 + taxa_juros_anual) ** periodo_anos)

print("\n--- Exemplo 2: Juros Compostos ---")
print("Capital Inicial: R$", capital_inicial)
print("Montante após 5 anos: R$", round(montante_acumulado, 2))
```

```python

# Exemplo 3: Logística e Lotes Indivisíveis (// e %)

demanda_total_unidades = 1435
capacidade_lote_padrao = 120

lotes_completos = demanda_total_unidades // capacidade_lote_padrao
unidades_excedentes = demanda_total_unidades % capacidade_lote_padrao

print("\n--- Exemplo 3: Planejamento Logístico ---")
print("Total de lotes completos:", lotes_completos)
print("Unidades restantes fora de lotes:", unidades_excedentes)
```

```python
# Exemplo 4: Taxa de Crescimento Anual Composta (CAGR) via Radiciação
# Formula: CAGR = (Valor_Final / Valor_Inicial)^(1/n) - 1

pib_inicial = 1.85  # Trilhões de reais no ano 0
pib_final = 2.42    # Trilhões de reais no ano 4
anos = 4

fator_crescimento = pib_final / pib_inicial
taxa_cagr = (fator_crescimento ** (1 / anos)) - 1
taxa_cagr_percentual = taxa_cagr * 100

print("\n--- Exemplo 4: Crescimento Econômico (CAGR) ---")
print("PIB Inicial: R$", pib_inicial, "trilhões")
print("PIB Final: R$", pib_final, "trilhões")
print("Taxa Anual Composta de Crescimento:", round(taxa_cagr_percentual, 2), "% ao ano")
```

## <span style="color:darkgreen"> 7)  Exercício de Fixação </span>

Considere um investidor que aplicou $R\$\;20.000,00$ em um fundo de investimento. Ao final de $3$ anos, o saldo total resgatado foi de $R\$\;28.500,00$.

Tarefas:

1) Armazene os valores do enunciado em variáveis descritivas.
2) Calcule a taxa de retorno nominal acumulada do período: $(\text{Valor Final} - \text{Valor Inicial}) / \text{Valor Inicial}$.
3) Calcule a taxa anual equivalente de juros por meio de radiciação.
4) Calcule a taxa contínua de retorno utilizando o logaritmo natural: $\ln(\text{Valor Final} / \text{Valor Inicial})$.
5) Exiba todas as respostas na tela de forma organizada.

```python
# Resolução

import math

# 1. Definição das variáveis
valor_inicial = 20000.0
valor_final = 28500.0
tempo_anos = 3

# 2. Retorno nominal acumulado
retorno_acumulado = (valor_final - valor_inicial) / valor_inicial

# 3. Taxa anualizada composta (radiciação)
taxa_anual_composta = ((valor_final / valor_inicial) ** (1 / tempo_anos)) - 1

# 4. Taxa contínua de retorno (logaritmo natural)
retorno_continuo = math.log(valor_final / valor_inicial)

# 5. Apresentação dos resultados
print("Retorno Nominal Acumulado:", round(retorno_acumulado * 100, 2), "%")
print("Taxa Anualizada Equivalente:", round(taxa_anual_composta * 100, 2), "% ao ano")
print("Retorno Contínuo Total:", round(retorno_continuo * 100, 2), "%")
```

```python
# Exercício 4

# 1. Variáveis macroeconômicas
pib_inicial_bilhoes = 420.0
pib_final_bilhoes = 595.0
periodos_anos = 6

# 2. Radiciação aplicada ao cálculo do CAGR
razao_crescimento = pib_final_bilhoes / pib_inicial_bilhoes
taxa_cagr = (razao_crescimento ** (1 / periodos_anos)) - 1
taxa_cagr_percentual = taxa_cagr * 100

# 3. Exibição dos resultados
print("--- RELATÓRIO DE CRESCIMENTO MACROECONÔMICO ---")
print("PIB 2018: US$", pib_inicial_bilhoes, "bilhões")
print("PIB 2024: US$", pib_final_bilhoes, "bilhões")
print("Taxa Composta de Crescimento Anual (CAGR):", round(taxa_cagr_percentual, 2), "% ao ano")
```

***Conceito Computacional:*** a precedência de operadores em Python atribui prioridade à potenciação sobre a divisão. Portanto, a expressão razao_crescimento ** 1 / periodos_anos calcularia incorretamente $(x^1) / 6$. O uso obrigatório de parênteses no expoente fracionário (1 / periodos_anos) garante que a divisão ocorra antes da elevação à potência, executando fielmente a raiz sexta ($\sqrt[6]{x}$).

***Interpretação Econômica:*** A taxa média geométrica de $5{,}98\%$ ao ano descreve o ritmo uniforme de expansão necessário para transformar $US\$\;420$ bilhões em $US\$\;595$ bilhões em $6$ anos, sendo uma métrica superior à média aritmética simples das taxas anuais, que superestima o crescimento acumulado.

```python
# Exercício 5

# 1. Parâmetros da tecnologia e fatores de produção
produtividade_total_fatores = 1.25
elasticidade_capital = 0.35
elasticidade_trabalho = 1.0 - elasticidade_capital  # Retornos constantes de escala (0.65)

capital = 10000.0
trabalho = 1500.0

# 2. Equação Cobb-Douglas: Y = A * (K^alpha) * (L^(1-alpha))
produto_total = produtividade_total_fatores * (capital ** elasticidade_capital) * (trabalho ** elasticidade_trabalho)

# 3. Produtividades médias
produtividade_media_trabalho = produto_total / trabalho
produtividade_media_capital = produto_total / capital

# 4. Saída formatada
print("--- ANÁLISE DE PRODUÇÃO COBB-DOUGLAS ---")
print("Parâmetro de Tecnologia (A):", produtividade_total_fatores)
print("Elasticidade do Capital (alpha):", elasticidade_capital)
print("Elasticidade do Trabalho (1-alpha):", elasticidade_trabalho)
print("Produto Total Agregado (Y):", round(produto_total, 2))
print("Produtividade Média do Trabalho (PMe_L):", round(produtividade_media_trabalho, 2), "unidades/trabalhador")
print("Produtividade Média do Capital (PMe_K):", round(produtividade_media_capital, 4), "unidades/capital")
```

***Conceito Computacional:*** demonstra o encadeamento de múltiplos termos com potenciação de expoentes em ponto flutuante (float). A legibilidade é mantida com nomes de variáveis bem estruturados e parênteses que isolam as operações individuais de potência antes da multiplicação final.

***Interpretação Econômica:*** como a soma dos expoentes é igual a $1{,}0$ ($\alpha + (1 - \alpha) = 1$), a função exibe Retornos Constantes de Escala (homogênea de grau 1). Se duplicarmos simultaneamente $K$ e $L$, o produto $Y$ dobrará na mesma proporção.

```python
# Exercício 6

import math

# 1. Declaração de preços e variância diária
preco_anterior = 920.0
preco_atual = 1085.0
variancia_diaria = 0.000484

# 2. Cálculo do Retorno Contínuo (Log-Return) com math.log
retorno_continuo = math.log(preco_atual / preco_anterior)
retorno_continuo_percentual = retorno_continuo * 100

# 3. Cálculo do Retorno Discreto Simples para fins comparativos
retorno_discreto = (preco_atual - preco_anterior) / preco_anterior
retorno_discreto_percentual = retorno_discreto * 100

# 4. Cálculo da Volatilidade Diária com math.sqrt
volatilidade_diaria = math.sqrt(variancia_diaria)
volatilidade_diaria_percentual = volatilidade_diaria * 100

# 5. Exibição dos resultados
print("--- DINÂMICA DE PREÇOS E RISCO DE COMMODITIES ---")
print("Preço Base (P_t-1): R$", preco_anterior)
print("Preço Final (P_t): R$", preco_atual)
print("Retorno Discreto Simples:", round(retorno_discreto_percentual, 2), "%")
print("Retorno Logarítmico Contínuo:", round(retorno_continuo_percentual, 2), "%")
print("Volatilidade Diária (Desvio-Padrão):", round(volatilidade_diaria_percentual, 2), "% ao dia")
```

***Conceito Computacional:*** a importação do módulo math expande as capacidades analíticas do Python sem necessidade de instalar pacotes externos. A função math.log(x) calcula por padrão o logaritmo natural ($\ln$). A função math.sqrt(x) executa o cálculo de raiz quadrada de modo altamente otimizado.

***Interpretação Econômica:*** o retorno contínuo ($16{,}50\%$) é ligeiramente inferior ao retorno discreto simples ($17{,}93\%$), refletindo a propriedade matemática de que $\ln(1 + R) \le R$. Na modelagem econométrica de ativos, o retorno logarítmico é adotado para garantir a simetria distributiva e a aditividade ao longo de múltiplos períodos temporais. A volatilidade diária de $2{,}20\%$ mensura a dispersão e o risco associado à commodity.
