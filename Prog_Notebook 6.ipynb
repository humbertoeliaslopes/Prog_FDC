### **<span style="color:darkgreen">1) Condições</span>**

As condições são fundamentais em um código porque permitem ao programa funcionar de forma diferente, de acordo com os valores informados pelo usuário.

Há três tipos fundamentais de estruturas condicionais:

a) IF (Se)

b) ELSE (Senão)

c) ELIF (Senão se)

Vejamos cada uma delas.

> **1.1) IF**

O IF é a estrutura básica das condições porque indica o que deve ser executado ou ignorado no código. Vejamos um exemplo.

```python
# Código para o usuário entrar com dois valores e somente o maior aparecer na tela

primeiro_valor = int(input("Primeiro valor: "))
segundo_valor = int(input("Segundo valor: "))

if primeiro_valor > segundo_valor: # Importante: lembre-se de digitar ":" ao final do IF.
    print("O primeiro valor é maior que o segundo!") # Atenção com a indentação!
if segundo_valor > primeiro_valor:
    print("O segundo valor é maior que o primeiro!") # Atenção com a indentação!
```

O código acima é muito simples, mas há um detalhe importante, destacado nas hashtags: a indentação.

Como sabemos, Python é uma linguagem que considera a indentação. No primeiro IF do código, queremos que, se a condição estabelecida por ele for atendida (for TRUE, em linguagem booleana), apareça a mensagem "O primeiro valor é maior que o segundo!". Essa ordem, portanto, faz parte do primeiro IF, e não do segundo. 

Para o Python entender isso, é preciso, então, recuar - ou indentar - a ordem de exibição da mensagem, vinculando-a ao primeiro IF. A mesma lógica serve para o segundo IF. 

Veja o que aconteceria se ignorássemos essa regra da semântica do Python:

```python
# Código para o usuário entrar com dois valores e somente o maior aparecer na tela, MAS DESRESPEITANDO AS REGRAS DE INTENDAÇÃO!

primeiro_valor = int(input("Primeiro valor: "))
segundo_valor = int(input("Segundo valor: "))

if primeiro_valor > segundo_valor:
print("O primeiro valor é maior que o segundo!") # Atenção com a indentação!
if segundo_valor > primeiro_valor:
print("O segundo valor é maior que o primeiro!") # Atenção com a indentação!
```

Como você pode verificar ao executar esse código, o Python exibe uma mensagem de erro porque não consegue entender o que você realmente quer exibir na tela.

Vejamos mais um código com IF, mas que respeita as regras da indentação.

```python
# Código para classificar se um carro é muito novo ou nada novo

from datetime import datetime  # Importa o módulo datetime para estabelecer o ano corrente

ano_atual = datetime.now().year  # Estabelece o ano corrente, permitindo rodar este código em qualquer época

ano_fabricacao = int(input("Digite o ano de fabricação do seu veículo: "))

if (ano_atual - ano_fabricacao) >= 3:
    print("O seu carro já não é muito novo!")
if (ano_atual - ano_fabricacao) < 3:
    print("O seu carro está muito novo!")
```

> **1.2) ELSE**

O ELSE é importante porque permite incluir todas as alternativas não abrangidas pelo IF. Por exemplo, no primeiro código do IF, não previmos a situação na qual os dois valores digitados são iguais. 

Vamos corrigir isso usando o ELSE.

```python
# Código para o usuário entrar com dois valores e somente o maior aparecer na tela. Agora usando o ELSE.

primeiro_valor = int(input("Primeiro valor: "))
segundo_valor = int(input("Segundo valor: "))

if primeiro_valor > segundo_valor: # Importante: lembre-se de digitar ":" ao final do IF.
    print("O primeiro valor é maior que o segundo!") 
if segundo_valor > primeiro_valor:
    print("O segundo valor é maior que o primeiro!") 
else:
    print("Ambos os valores são iguais")
```

> **IMPORTANTE: ESTRUTURAS ANINHADAS**

Muitas vezes, precisamos utilizar IF e ELSE dentro de outros para obtermos o resultado desejado. Esse procedimento é denominado uso de estruturas aninhadas.

Um bom exemplo de como essas estruturas funcionam é o código de classificação dos carros de acordo com a idade, que desenvolvemos anteriormente.

Vamos mudar esse código, supondo as seguintes situações:

a) Se o carro tiver mais de 3 anos, será considerado "Não muito novo".

b) Se tiver menos de 3 anos, será considerado "Muito novo.""

c) Se tiver exatamente 3 anos, será considerado "Mais ou menos novo."

Usemos o ELSE para fazer as mudanças.

```python
# Código para classificar um carro, usando o ELSE

from datetime import datetime

ano_atual = datetime.now().year
ano_fabricacao = int(input("Digite o ano de fabricação do seu veículo: "))
idade_carro = ano_atual - ano_fabricacao

if idade_carro > 3:
    print("O seu carro já não é muito novo!")
else:
    # Se chegou aqui, o carro tem 3 anos ou menos
    if idade_carro < 3: # IF aninhado
        print("O seu carro está muito novo!")
    else: # ELSE aninhado
        print("O seu carro é mais ou menos novo!")
```

Vejamos outro exemplo. Suponhamos que você queira criar um código que estime o valor que um usuário pagará em sua conta de telefone. Mas há três faixas de preço:

a) Se o usuário gastar menos de 200 minutos, o preço do minuto será R$ 0,20. 

b) Do contrário, se ele gastar menos de 400 minutos, o preço será de R$ 0,18. 

c) Qualquer outra situação além dessas implicará uma cobrança de R$ 0,15 por minuto.

Observe que podemos criar um código no qual há três situações:

a) O usuário gasta menos de 200 minutos, pagando R$ 0,20 por minuto.

b) O usuário gasta menos de 400 minutos e, pelo menos, 200 minutos, pagando R$ 0,18 por minuto.

c) O usuário gasta mais que 400 minutos, pagando R$ 0,15 por minuto. 

Claramente, a estratégia de precificação da companhia telefônica estimula seus usuários a usarem mais minutos. 

Para incorporar essas três situações por meio de estruturas aninhadas, criamos o seguinte código:

```python
# Código para calcular o valor da conta de telefone, a partir de três faixas de preço

minutos = int(input("Digite a quantidade de minutos utilizados: "))

if minutos < 200:
    valor = minutos * 0.20
else:
    if minutos < 400:
        valor = minutos * 0.18
    else:
        valor = minutos * 0.15

print(f"O valor da sua conta de telefone é: R$ {valor:.2f}")
```

Vejamos outro exemplo de uso das estruturas aninhadas. Suponhamos que uma loja precise de um programa que mostre o preço de cinco diferentes categorias de produto. Os dados são os seguintes:


| Categoria | Preço |
| :---: | :---: |
| A | 10,00 |
| B | 18,00 |
| C | 23,00 |
| D | 26,00 |
| E | 31,00 |


A partir desses dados, podemos criar o seguinte código em Python:

```python
# Código para identificar o preço do produto, de acordo com sua categoria

categoria = input("Digite a categoria do produto (A, B ou C): ").upper() # .upper() transforma a entrada em maiúscula   

if categoria == "A":
    preco = 10.00
else:
    if categoria == "B":
        preco = 18.00
    else:
        if categoria == "C":
            preco = 23.00
        else:
            if categoria == "D":
                preco = 26.00
            else:
                if categoria == "E":
                    preco = 31.00
                else:
                    print("Categoria inválida! Digite A, B, C, D ou E.")
                    preco = 0.00 # Preço zero para categoria inválida
print(f"O preço do produto da categoria {categoria} é: R$ {preco:.2f}")                        
```

> **1.3) ELIF**

O comando ELIF (ELSE IF) é uma boa alternativa para evitarmos múltiplos IF aninhados, o que simplifica enormemente o código. Mais especificamente, ELIF substitui um par ELSE IF, sem criar outro nível de estrutura, demandando indentação. 

Vejamos como ELIF funciona reescrevendo o código anterior.

```python
# Código para identificar o preço do produto, de acordo com sua categoria e usando ELIF

categoria = input("Digite a categoria do produto (A, B ou C): ").upper() # .upper() transforma a entrada em maiúscula   

if categoria == "A":
    preco = 10.00
elif categoria == "B":
    preco = 18.00
elif categoria == "C":
    preco = 23.00
elif categoria == "D":
    preco = 26.00
elif categoria == "E":
    preco = 31.00
else:
    print("Categoria inválida! Digite A, B, C, D ou E.")
    preco = 0.00 # Preço zero para categoria inválida
print(f"O preço do produto da categoria {categoria} é: R$ {preco:.2f}")     
```

Observe como o código ficou BEM mais simples!

### **<span style="color:darkgreen">2) Repetições</span>**

As repetições são usadas para executar a mesma parte de um código diversas vezes, normalmente dependendo de uma condição. O principal comando de repetição é o WHILE (ENQUANTO, no pseudocódigo).

> **2.1) WHILE**

O comando WHILE faz com que o código repita uma instrução até que a condição mude, sustando a execução. Vejamos um exemplo simples.

Vamos criar um código em que a variável var_teste receberá o valor inicial de 0 (zero). Depois, estabeleceremos que o código imprimirá o valor armanzenado em var_teste enquanto (WHILE) a variável receber valor menor ou igual a 5. 

O código ficará assim:

```python
# Código para testar a função WHILE

var_teste = 0 # A variável recebe o valor 0, que é menor que 5, então o WHILE será executado
while var_teste <= 5: # Enquanto a variável for menor ou igual a 5, o WHILE será executado
    print(f"O valor da variável é: {var_teste}")
    var_teste += 1 # A variável é incrementada em 1 a cada execução do WHILE, até que seja maior que 5, quando o WHILE será encerrado.
```

Um detalhe importante neste código: usamos o comando += para adicionar 1 a cada ciclo de repetição do código. Essa é uma maneira mais direta de especificar essa ação no Python. Alternativamente, você poderia ter usado o comando mais tradicional:

**var_teste = var_teste + 1**

Veja que a alternativa que usamos economiza caracteres, tornando o código mais objetivo.

Podemos expandir as aplicações das repetições por meio de duas estruturas auxiliares, que são os **contadores** e os **acumuladores**.

> **2.2) Contadores**

Os contadores estabelecem um limite para as repetições em um código. Assim, eles automatizam o número de ciclos que será rodado, facilitando o controle do usuário.

Para entendermos isso melhor, suponhamos que quiséssemos mudar o código anterior para ele registrar valores de var_teste até 1000. Sem usar um contador, teríamos o seguinte resultado:

```python
# Código para testar a função WHILE sem contador, apenas com a condição de parada

var_teste = 0 # A variável recebe o valor 0, que é menor que 5, então o WHILE será executado
while var_teste <= 5: # Enquanto a variável for menor ou igual a 5, o WHILE será executado
    print(f"O valor da variável é: {var_teste}")
    var_teste += 1 # A variável é incrementada em 1 a cada execução do WHILE, até que seja maior que 5, quando o WHILE será encerrado.
```

Agora suponhaos que desejamos parar a execução do código quando var_teste receber um valor como 35, por exemplo. Ora, o código está programado para rodar até a variável receber um valor bem maior que esse. Logo, precisaremos modificar nossa programação para que ele possa ser interrompido no limite que desejamos. 

Para isso, vamos adicionar um contador à repetição:

```python
# Código para testar a função WHILE, com contador, apenas com a condição de parada

var_contador = int(input("Digite o valor máximo para o contador: ")) # A variável recebe o valor digitado pelo usuário, que será o limite do WHILE
var_teste = 0 
while var_teste <= var_contador: 
    print(f"O valor da variável é: {var_teste}")
    var_teste += 1 
```

Com essa mudança, o código rodará até o limite estabelecido por nós. 

> **2.3) Acumuladores**

Os contadores são muito úteis, mas têm uma limitação: são efetivados por valores fixos. Mas, e se precisarmos estabelecer um limite variável de ciclos no código? Neste caso, recorreremos aos **acumuladores**.

Um recurso complementar muito útil a eles são os laços (ou *loops*). Eles são implementados por meio do comando **FOR**, seguido do índice "*i*", que representa o número de execuções do código. Além disso, podemos usar a função *range()* para determinar o número de execuções, completando as informações necessárias para o laço.

Como exemplo, vamos criar um código que calcule a soma de 10 números digitados pelo usuário.

```python
# Código para calcular a soma de 10 números digitados pelo usuário.

acumulador = 0 # A variável acumulador recebe o valor 0, que será somado aos números digitados pelo usuário
for i in range(10): # O laço FOR será executado 10 vezes, de 0 a 9, totalizando 10 execuções. O índice "i" representa cada execução.
    numero = int(input("Digite um número: "))
    acumulador += numero
print(f"A soma dos números é: {acumulador}")
```

Vamos modificar esse código, para que o acumulador nos leve a calcular a média aritmética de 15 números.

```python
# Código para calcular a média aritmética de 15 números digitados pelo usuário.

acumulador = 0 # A variável acumulador recebe o valor 0, que será somado aos números digitados pelo usuário
for i in range(15): # O laço FOR será executado 15 vezes, de 0 a 14, totalizando 15 execuções. O índice "i" representa cada execução.
    numero = int(input("Digite um número: "))
    acumulador += numero
print(f"A média dos números é: {acumulador / 15}")
```

> **2.4) Interrompendo a Repetição**

Algumas vezes, contadores e acumuladores não são tão úteis quando precisamos interromper a execução do código. Justamente por isso, o Python nos oferece o comando **BREAK**.

Vejamos um exemplo.

```python
# Código que usa o comando BREAK para interromper a execução.

acumulador = 0
while True: # Laço infinito, que só será interrompido pelo comando BREAK
    numero = int(input("Digite um número (ou 0 para sair): "))
    if numero == 0:
        break # Interrompe o laço quando o usuário digita 0
    acumulador += numero
print(f"A soma dos números digitados é: {acumulador}")
```

Vamos trabalhar em um exemplo mais complexo. Nossa tarefa é escrever um código para controlar as vendas de produtos em uma loja. O usuário deve digitar o código do produto e a quantidade vendida. A tabela a seguir apresenta os preços de cada produto.

| Código | Preço |
| :---: | :---: |
| 1 | 15,50 |
| 2 | 33,20 |
| 3 | 47,80 |
| 5 | 21,90 |
| 9 | 76,39 |

O código deve exibir o total das vendas depois que o usuário digitar "0". Além disso, qualquer outro código de produto deve resultar na mensagem de erro: "Código Inválido". 


```python
# Código para calcular o total das vendas de uma loja.

# Inicializa a variável para acumular o valor total das vendas
total_vendas = 0.0

print("--- Sistema de Controle de Vendas ---")
print("Digite '0' no código do produto para encerrar a venda.\n") # \n é usado para pular uma linha

while True: # Função que ativa o laço infinito, que só será interrompido pelo comando BREAK
    # Solicita o código do produto ao usuário
    codigo = int(input("Digite o código do produto: "))
    
    # Condição de parada do loop
    if codigo == 0:
        break
        
    # Verifica o preço com base no código informado
    if codigo == 1:
        preco = 15.50
    elif codigo == 2:
        preco = 33.20
    elif codigo == 3:
        preco = 47.80
    elif codigo == 5:
        preco = 21.90
    elif codigo == 9:
        preco = 76.39
    else:
        # Se o código não estiver na tabela, exibe erro e reinicia o loop
        print("Código Inválido\n")
        continue

    # Se o código for válido, solicita a quantidade vendida
    quantidade = int(input(f"Digite a quantidade vendida do produto {codigo}: "))
    
    # Calcula o subtotal e adiciona ao total geral
    total_vendas += preco * quantidade
    print(f"Subtotal atual: R$ {total_vendas:.2f}\n")

# Exibe o resultado final após a digitação do código 0
print("-" * 35)
print(f"VALOR TOTAL DAS VENDAS: R$ {total_vendas:.2f}")
print("-" * 35)
```

> **2.5) Repetições Aninhadas**

O recurso de aninhamento de instruções também pode ser usado em repetições. Vejamos o exemplo de um código que cria uma tabuada dos números de 1 a 10:

```python
# Tabuada de 1 a 10, usando dois loops FOR aninhados

# Loop externo: passa por cada número de 1 a 10
for i in range(1, 11):
    print(f"--- Tabuada do {i} ---")
    
    # Loop interno: multiplica o número atual por todos os valores de 1 a 10
    for j in range(1, 11):
        resultado = i * j
        print(f"{i} x {j} = {resultado}")
        
    print()  # Linha em branco para separar as tabelas
```

### **<span style="color:darkgreen">3) F-Strings</span>**

As denominadas *F-Strings* foram introduzidas na versão 3.6 do Python para que o usuário pudesse substituir o valor de uma variável ou expressão em uma string. Vejamos alguns exemplos.

```python
var_teste = "sobre F-Strings!"
print(f"Este é um Jupyter Notebook {var_teste}")
```

```python
var_teste2 = "Programação para o curso de Economia"
print(f"Disciplina: {var_teste2}")
```

As F-Strings também são úteis quando precisamos especificar o número de caracteres ou de casas decimais armazenados em uma variável. Exemplos:

```python
preco = 13.45
f"Preço do produto: R$ {preco:.2f}" #.2f limita o número de casas decimais a 2, e o "f" indica que é uma f-string
```

```python
altura = 1.73
print(f"Altura do usuário: {altura} metros")
```

As F-Strings também permitem alinhar valores à direita, esquerda ou ao centro, usando os símbolos >, < e ^, respectivamente. Exemplos:

```python
# Alinhamento à esquerda, com duas casas decimais, usando f-string

altura = 1.57
print(f"Altura do usuário: {altura:<10.2f} metros") # O valor 10 indica o tamanho do campo reservado para o valor, e o "<" indica que o valor será alinhado à esquerda.
```

```python
# Alinhamento à direita, com duas casas decimais, usando f-string

altura = 1.57
print(f"Altura do usuário: {altura:>10.2f} metros")
```

```python
# Alinhamento ao centro, com duas casas decimais, usando f-string

altura = 1.57
print(f"Altura do usuário: {altura:^10.2f} metros")
```
