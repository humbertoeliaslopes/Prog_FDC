## <span style="color:darkgreen"> 1) Nomes de Variáveis </span>

Em Python, os nomes das variáveis devem se iniciar com uma letra (maiúscula ou minúscula) ou com o sublinhado (_). Após o primeiro caractere, os nomes podem conter letras, números e o sublinhado novamente.

Em versões anteriores do Python os nomes das variáveis jamais poderiam ser acentuados. A partir da versão 3, porém, a linguagem passou a usar um conjunto de caracteres denominado UTF-8, tornando-se capaz de representar praticamente todas as letras dos alfabetos conhecidos.

Apesar disso, é fortemente recomendado que você evite acentuação em nomes de variáveis.

### **<span style="color:darkgreen">2.1.1) Arredondamento de Variáveis Quantitativas (ou Numéricas)</span>**

Em Python, o arredondamento de valores é realizado mediante a função round(). Veja os exemplos.

### **<span style="color:green">Na Prática 4.3</span>**: Arredondando Valores de Variáveis Quantitativas


Armazene os resultados das operações a seguir em variáveis, arredondando seus valores conforme indicado. Não se esqueça de verificar o tipo de variável antes e depois do arredondamento.

a) 4,123 / 3,907 (divisão módulo, quatro casas decimais)

b) log2 (4,567) ** 5 (três casas decimais)

c) raiz quarta de 3,45 - 12 * 0,34 (cinco casas decimais)

d) 3,12 / 6,09 (divisão convencional, duas casas decimais)

e) raiz quadrada de 1,234 * 0,033 (seis casas decimais)

### **<span style="color:green">Resolução</span>**

```python
# Letra a

op_1 = round(4.123 / 3.907, 4)
print(op_1)
type(op_1)
```

```python
# Letra b

from math import log2
op_2 = round((log2(4.567)) ** 5, 3)
print(op_2)
type(op_2)
```

```python
# Letra c

# Atenção: o resultado de 3,45 -12 * 0,34 é negativo, então a função abs() foi utilizada para obter o valor absoluto antes de calcular a raiz quarta.

op_3 = round(abs(3.45 - 12 * 0.34) ** (1 / 4), 5)
print(op_3)
type(op_3)
```

```python
# Letra d

op_4 = round(3.12 / 6.09, 2)
print(op_4)
type(op_4)  
```

```python
# Letra e

# Como o cálculo envolve uma raiz quadrada, vamos usar a função sqrt() da biblioteca math para calculá-la.

from math import sqrt
op_5 = round(sqrt(1.234 * 0.033), 6)
print(op_5)
type(op_5)
```

### **<span style="color:darkgreen">2.2) Variáveis Lógicas</span>**

Quando escrevemos códigos, frequentemente queremos armazenar um conteúdo do tipo verdadeiro ou falso em uma variável. Nessas situações, devemos trabalhar com um tipo de variável chamado de lógico ou booleano.

Para operacionalizar esse tipo de variável em Python, escrevemos **True** para verdadeiro e **False** para falso. As iniciais são sempre maiúsculas.

Além disso, para realizarmos as comparações lógicas com variáveis booleanas, devemos empregar os operadores relacionais. Em Python eles são os seguintes:

| Operador | Operação | Símbolo Matemático |
| :---: | :---: | :---: |
| == | Igualdade | $= \ $ |
| > | Maior que | $>$ |
| < | Menor que   | $<$ |
| != | Diferente   | $\neq$ |
| >= | Maior ou igual a | $\geq$ |
| <= | Menor ou igual a   | $\leq$ |
| not | Não   | - |
| and | E   | - |
| or | Ou   | - |

### **<span style="color:green">Na Prática 4.5</span>**: Trabalhando com Variáveis Booleanas - Parte II


Usando os operadores booleanos adequados, escreva o algoritmo para testar as seguintes condições:

a) O financiamento de um automóvel via Crédito Direto ao Consumidor (CDC) será concedido somente se o pleiteante tiver mais de 18 anos e ganhar um salário mensal igual ou superior a R$ 5.000. Teste esta condição sob diferentes contextos.

b) A economia de um país estará em recessão se apresentar dois trimestres consecutivos de declínio no PIB real. Crie diferentes cenários para testar seu algoritmo.

c) Um cidadão poderá pleitear determinado benefício de uma associação privada se tiver mais de 5 anos de afiliação ou idade de 50 anos, pelo menos. Use diferentes valores para testar a consistência do seu algoritmo.

### **<span style="color:green">Resolução</span>**

```python
# Letra a

# Entrada de dados
idade = 25
salario_mensal = 5200.00

# Regra com operadores booleanos
# Mais de 18 anos E salário igual ou superior a 5000
pode_financiar = (idade > 18) and (salario_mensal >= 5000.00)

# Verificação
if pode_financiar:
    print("Financiamento CDC APROVADO.")
else:
    print("Financiamento CDC REPROVADO. Requisitos mínimos não atingidos.")
```

```python
# Letra b

# Entrada de dados (Variação percentual do PIB real nos dois últimos trimestres)
pib_trimestre_1 = -0.2  # queda de 0.2%
pib_trimestre_2 = -0.5  # queda de 0.5%

# Regra com operadores booleanos
# Primeiro trimestre negativo E segundo trimestre negativo
em_recessao = (pib_trimestre_1 < 0) and (pib_trimestre_2 < 0)

# Verificação
if em_recessao:
    print("A economia do país está oficialmente em RECESSÃO TÉCNICA.")
else:
    print("A economia do país NÃO está em recessão.")
```

```python
# Entrada de dados
anos_afiliacao = 3
idade_cidadao = 52

# Regra com operadores booleanos
# Mais de 5 anos de afiliação OU idade de 50 anos pelo menos (maior ou igual)
elegivel_ao_beneficio = (anos_afiliacao > 5) or (idade_cidadao >= 50)

# Verificação
if elegivel_ao_beneficio:
    print("Benefício CONCEDIDO ao associado.")
else:
    print("Benefício NEGADO. O associado não cumpre nenhum dos requisitos mínimos.")
```
