### **<span style="color:darkgreen">1) O que é uma função?</span>**

Uma função é um bloco de código que recebe um nome e pode ser executado sempre que precisarmos dele. Você já usou várias funções prontas do Python nos Notebooks anteriores, como **print()**, **input()** e **len()**.

A vantagem de criar nossas próprias funções é evitar a repetição de código. Se um cálculo aparece muitas vezes em um programa, nós o escrevemos uma única vez dentro de uma função e depois apenas a chamamos pelo nome.

> **1.1) O comando DEF**

Para criar uma função, usamos o comando **DEF** (abreviação de *define*, ou seja, definir). A estrutura básica é a seguinte:

```python
def nome_da_funcao():
    instruções da função
```

Observe três detalhes importantes:

a) Depois do nome da função, colocamos parênteses.

b) A linha do DEF termina com ":", assim como acontece no IF e no WHILE.

c) As instruções que pertencem à função ficam indentadas.

Vejamos um exemplo.

````python
# Código para criar uma função que exibe o cabeçalho de um relatório econômico

def exibir_cabecalho(): # Cria (define) a função. Atenção aos parênteses e aos dois pontos!
    print("=" * 40)
    print("RELATÓRIO DE INDICADORES ECONÔMICOS") # Atenção com a indentação!
    print("=" * 40)
````

Ao executar a célula acima, nada aparece na tela. Isso acontece porque apenas **criamos** a função. Para que ela seja executada, precisamos **chamá-la** pelo nome, sempre com os parênteses.

````python
# Chama a função criada acima

exibir_cabecalho()
````

Podemos chamar a mesma função quantas vezes quisermos, sem precisar reescrever as instruções.

````python
# Chama a função duas vezes

exibir_cabecalho()
print("Indicador: inflação mensal")
exibir_cabecalho()
print("Indicador: taxa de desemprego")
````

### **<span style="color:darkgreen">2) Parâmetros</span>**

Os parâmetros são as informações que enviamos para dentro da função. Eles são escritos entre os parênteses, logo após o nome da função, e funcionam como variáveis que recebem os valores informados no momento da chamada.

> **2.1) Função com um parâmetro**

Vamos criar uma função que calcula o valor de um produto após um reajuste de 10%. O preço do produto será o parâmetro.

````python
# Código para calcular o novo preço de um produto após um reajuste de 10%

def reajustar_preco(preco): # "preco" é o parâmetro da função
    novo_preco = preco * 1.10
    print(f"Preço antigo: R$ {preco:.2f} | Novo preço: R$ {novo_preco:.2f}")

reajustar_preco(50.00)  # O valor 50.00 é enviado para o parâmetro "preco"
reajustar_preco(120.00) # Agora o parâmetro "preco" recebe 120.00
````

> **2.2) Função com vários parâmetros**

Uma função pode receber mais de um parâmetro. Nesse caso, eles são separados por vírgulas. 

Como exemplo, vamos calcular o PIB *per capita* de um município, que é o PIB dividido pela população. Os valores usados são hipotéticos.

````python
# Código para calcular o PIB per capita de um município

def calcular_pib_per_capita(pib, populacao): # Dois parâmetros, separados por vírgula
    pib_per_capita = pib / populacao
    print(f"PIB per capita: R$ {pib_per_capita:.2f}")

calcular_pib_per_capita(2500000000, 50000) # O primeiro valor vai para "pib" e o segundo vai para "populacao"
````

Atenção: a ordem dos valores importa! O primeiro valor informado vai para o primeiro parâmetro, o segundo valor vai para o segundo parâmetro, e assim por diante.

Se quisermos evitar confusão, podemos informar o nome de cada parâmetro na chamada da função. Assim, a ordem deixa de ser importante.

````python
# Chamando a função com o nome dos parâmetros

calcular_pib_per_capita(populacao=50000, pib=2500000000) # O resultado é o mesmo, mesmo com a ordem trocada
````

> **2.3) Parâmetros com valor padrão**

Também podemos definir um valor padrão para um parâmetro. Se o usuário não informar esse valor na chamada, a função usará o valor padrão.

Vejamos um exemplo com juros simples, cuja fórmula é:

**J = C × i × t**

em que **C** é o capital aplicado, **i** é a taxa de juros por período e **t** é o número de períodos. Vamos supor que, na maior parte das aplicações, o prazo seja de 12 meses.

````python
# Código para calcular juros simples, com prazo padrão de 12 meses

def calcular_juros_simples(capital, taxa, tempo=12): # Se "tempo" não for informado, ele valerá 12
    juros = capital * taxa * tempo
    print(f"Capital: R$ {capital:.2f} | Taxa: {taxa * 100:.1f}% ao mês | Prazo: {tempo} meses | Juros: R$ {juros:.2f}")

calcular_juros_simples(1000, 0.01)    # Usa o prazo padrão de 12 meses
calcular_juros_simples(1000, 0.01, 6) # Informa um prazo de 6 meses
````

### **<span style="color:darkgreen">3) Retornando valores com RETURN</span>**

Até agora, nossas funções apenas exibiram os resultados na tela com o **print()**. Muitas vezes, porém, precisamos **guardar** o resultado de uma função em uma variável para usá-lo em outros cálculos. Para isso, usamos o comando **RETURN**.

O RETURN devolve um valor para quem chamou a função e encerra a execução dela.

> **3.1) A diferença entre PRINT e RETURN**

Vamos criar uma função que calcula o lucro de uma empresa, que é a receita menos o custo.

````python
# Código para calcular o lucro de uma empresa usando RETURN

def calcular_lucro(receita, custo):
    lucro = receita - custo
    return lucro # Devolve o valor do lucro para quem chamou a função

resultado = calcular_lucro(80000, 65000) # O valor devolvido é guardado na variável "resultado"
print(f"Lucro da empresa: R$ {resultado:.2f}")
````

Como o resultado foi guardado na variável, podemos usá-lo em novos cálculos. Por exemplo, podemos calcular a margem de lucro, que é o lucro dividido pela receita.

````python
# Código para calcular a margem de lucro, aproveitando a função calcular_lucro()

receita = 80000
lucro = calcular_lucro(receita, 65000)
margem = lucro / receita * 100

print(f"Lucro: R$ {lucro:.2f}")
print(f"Margem de lucro: {margem:.1f}%")
````

Agora veja o que acontece se a função apenas exibir o resultado com o **print()**, sem usar o RETURN.

````python
# Código para calcular o lucro SEM usar RETURN

def mostrar_lucro(receita, custo):
    lucro = receita - custo
    print(f"Lucro: R$ {lucro:.2f}") # Apenas exibe o valor, mas não o devolve

resultado = mostrar_lucro(80000, 65000)
print(resultado) # Como a função não devolveu nada, a variável recebe o valor None (vazio)
````

Perceba que a variável **resultado** recebeu o valor **None**, que significa "nada" em Python. Por isso, sempre que precisarmos reaproveitar o resultado de uma função, devemos usar o RETURN.

> **3.2) Mais exemplos com RETURN**

Vamos calcular a taxa de inflação entre dois períodos a partir de um índice de preços. A fórmula é:

**Inflação (%) = (Índice final ÷ Índice inicial − 1) × 100**

Os valores do índice usados abaixo são hipotéticos.

````python
# Código para calcular a taxa de inflação entre dois períodos

def calcular_inflacao(indice_inicial, indice_final):
    inflacao = (indice_final / indice_inicial - 1) * 100
    return inflacao

inflacao_ano = calcular_inflacao(6500.00, 6780.00)
print(f"A inflação no período foi de {inflacao_ano:.2f}%")
````

Uma função também pode conter estruturas condicionais, como as que estudamos no Notebook 6. 

No exemplo a seguir, calculamos a elasticidade-preço da demanda, que é a variação percentual da quantidade demandada dividida pela variação percentual do preço. Em seguida, outra função classifica o resultado:

a) Se o valor absoluto da elasticidade for maior que 1, a demanda é **elástica**.

b) Se for menor que 1, a demanda é **inelástica**.

c) Se for igual a 1, a demanda tem **elasticidade unitária**.

````python
# Código para calcular e classificar a elasticidade-preço da demanda

def calcular_elasticidade(var_quantidade, var_preco): # Variações em porcentagem
    return var_quantidade / var_preco # O RETURN pode devolver diretamente o resultado de um cálculo

def classificar_elasticidade(elasticidade):
    if abs(elasticidade) > 1: # abs() retorna o valor absoluto (sem o sinal negativo)
        return "elástica"
    elif abs(elasticidade) < 1:
        return "inelástica"
    else:
        return "de elasticidade unitária"

# O preço subiu 10% e a quantidade demandada caiu 20%
e = calcular_elasticidade(-20, 10)
print(f"Elasticidade: {e:.2f}. A demanda é {classificar_elasticidade(e)}.")

# O preço subiu 10% e a quantidade demandada caiu 4%
e = calcular_elasticidade(-4, 10)
print(f"Elasticidade: {e:.2f}. A demanda é {classificar_elasticidade(e)}.")
````

### **<span style="color:darkgreen">4) Escopo: variáveis locais e globais</span>**

O escopo de uma variável é a parte do código em que ela existe e pode ser usada. Em Python, há dois tipos principais de escopo:

a) **Local**: variáveis criadas **dentro** de uma função. Elas só existem enquanto a função está sendo executada.

b) **Global**: variáveis criadas **fora** de qualquer função, no corpo principal do código. Elas podem ser lidas em qualquer parte do programa.

> **4.1) Variáveis locais**

Vamos criar uma função que calcula o valor do imposto sobre uma venda. A variável **imposto** é criada dentro da função e, portanto, é local.

````python
# Código para calcular o imposto sobre uma venda, com alíquota de 18%

def calcular_imposto(valor_venda):
    imposto = valor_venda * 0.18 # "imposto" é uma variável LOCAL
    print(f"Imposto dentro da função: R$ {imposto:.2f}")

calcular_imposto(500.00)
````

Agora, veja o que acontece quando tentamos acessar a variável **imposto** fora da função.

````python
# Tentando acessar uma variável local fora da função

print(imposto) # Isso causará um erro, pois "imposto" só existe dentro da função
````

O Python exibe a mensagem *NameError*, informando que a variável **imposto** não foi definida. Isso acontece porque ela deixou de existir assim que a função terminou. Se precisarmos desse valor fora da função, devemos usar o RETURN, como vimos na seção 3.

> **4.2) Variáveis globais**

Uma variável criada fora de qualquer função é global. Ela pode ser **lida** dentro das funções. No exemplo a seguir, a taxa básica de juros é definida uma única vez, no corpo principal do código, e usada dentro da função. O valor da taxa é hipotético.

````python
# Código que usa uma variável global dentro de uma função

taxa_juros = 0.105 # Variável GLOBAL: taxa de juros de 10,5% ao ano

def calcular_rendimento(capital):
    rendimento = capital * taxa_juros # A função consegue LER a variável global
    return rendimento

print(f"Rendimento anual de R$ 1000.00: R$ {calcular_rendimento(1000):.2f}")
print(f"Rendimento anual de R$ 5000.00: R$ {calcular_rendimento(5000):.2f}")
````

A vantagem desse recurso é que, se a taxa de juros mudar, basta alterar o valor em um único lugar do código.

> **4.3) A palavra-chave GLOBAL**

Uma função consegue **ler** uma variável global, mas não consegue **alterá-la** diretamente. Vejamos o que acontece quando tentamos atualizar o total de vendas de uma loja dentro de uma função.

````python
# Tentando alterar uma variável global dentro de uma função, SEM usar GLOBAL

total_vendas = 0 # Variável global que acumula as vendas da loja

def registrar_venda(valor):
    total_vendas = total_vendas + valor # Isso causará um erro!

registrar_venda(150.00)
````

O Python exibe a mensagem *UnboundLocalError*. Isso acontece porque, ao encontrar **total_vendas = ...** dentro da função, o Python entende que estamos criando uma nova variável **local** com esse nome. Como essa variável local ainda não tem valor, a soma não pode ser feita.

Para informar ao Python que queremos alterar a variável **global**, usamos a palavra-chave **GLOBAL** no início da função.

````python
# Alterando uma variável global dentro de uma função, COM o uso de GLOBAL

total_vendas = 0

def registrar_venda(valor):
    global total_vendas # Avisa ao Python que vamos alterar a variável global
    total_vendas = total_vendas + valor

registrar_venda(150.00)
registrar_venda(320.50)
registrar_venda(89.90)

print(f"Total de vendas do dia: R$ {total_vendas:.2f}")
````

> **4.4) Uma boa prática**

Embora a palavra-chave GLOBAL funcione, os programadores costumam evitá-la. Quando muitas funções alteram as mesmas variáveis globais, fica difícil descobrir onde um valor foi modificado, e os erros se tornam mais difíceis de encontrar.

Na maioria das vezes, a alternativa mais segura é enviar o valor por meio de um parâmetro e receber o resultado com o RETURN. Veja o mesmo exemplo reescrito dessa forma.

````python
# Mesmo exemplo das vendas, agora usando parâmetro e RETURN, sem GLOBAL

def registrar_venda(total, valor):
    return total + valor # Devolve o novo total

total_vendas = 0
total_vendas = registrar_venda(total_vendas, 150.00)
total_vendas = registrar_venda(total_vendas, 320.50)
total_vendas = registrar_venda(total_vendas, 89.90)

print(f"Total de vendas do dia: R$ {total_vendas:.2f}")
````

> **4.5) Funções locais**

Assim como as variáveis, as funções também podem ser locais. Uma função criada **dentro** de outra função só pode ser usada pela função que a contém.

No exemplo a seguir, a função **calcular_ir()** calcula o imposto de renda de uma empresa, com alíquota hipotética de 15%, e só é usada dentro de **calcular_lucro_liquido()**.

````python
# Código com uma função local, criada dentro de outra função

def calcular_lucro_liquido(receita, custo):

    def calcular_ir(lucro): # Função LOCAL: só existe dentro de calcular_lucro_liquido()
        return lucro * 0.15

    lucro_bruto = receita - custo
    lucro_liquido = lucro_bruto - calcular_ir(lucro_bruto)
    return lucro_liquido

print(f"Lucro líquido: R$ {calcular_lucro_liquido(80000, 65000):.2f}")
````

Se tentarmos chamar a função local fora da função principal, o Python exibirá um erro, pelo mesmo motivo que vimos com as variáveis locais.

````python
# Tentando chamar a função local fora da função principal

calcular_ir(15000) # Isso causará um erro, pois calcular_ir() só existe dentro de calcular_lucro_liquido()
````

Já as funções que criamos no corpo principal do código, como **calcular_lucro()** e **calcular_inflacao()**, são globais e podem ser chamadas em qualquer parte do programa.
