### **<span style="color:darkgreen">Etapa 4) Quem puxou a inflação? Contribuição dos grupos do IPCA (0,5 ponto)</span>**

**Objetivo:** usar os dados do SIDRA para decompor a inflação dos últimos 12 meses da amostra entre os grupos de produtos e serviços.

A contribuição de um grupo para a variação do IPCA em um mês, em pontos percentuais, é aproximada por:

$$\text{contribuição}_{g,t} = \frac{\text{peso}_{g,t} \times \text{variação}_{g,t}}{100}$$

O modelo é:

$$\pi_t^{12m} = \beta_0 + \beta_1 \, u_t + \varepsilon_t$$

em que $\pi_t^{12m}$ é o IPCA acumulado em 12 meses e $u_t$ é a taxa de desocupação. Pelo método de mínimos quadrados ordinários (MQO), os coeficientes são:
