Neste Jupyter Notebook, você aprenderá os princípios que regem a semântica do Python. Ao contrário da sintaxe, a semântica de uma linguagem diz respeito ao **SIGNIFICADO** dos comandos.

### **<span style="color:green">PRINCÍPIO 1</span>**:  as variáveis em Python são ***PONTEIROS***. 

Diferente de linguagens como C ou C++, onde uma variável é uma "caixa" na memória que guarda um valor, em Python, uma variável é apenas uma etiqueta (um ponteiro) que aponta para um objeto na memória.

```python
# EXEMPLO 1.1

# Criando uma lista
lista_a = [1, 2, 3]

# Você não está copiando a lista, está apenas criando outra etiqueta para a MESMA lista
lista_b = lista_a

# Se alterarmos a lista_b...
lista_b.append(4)

# A lista_a TAMBÉM foi alterada!
print(lista_a)  # Resultado: [1, 2, 3, 4]
```

Isso acontece justamente porque tanto lista_a quanto lista_b são ponteiros (etiquetas) que apontam para o mesmíssimo endereço de memória onde a lista real está guardada.

```python
# Exemplo 1.2

x = 1 # x é um valor inteiro
x = 'hello' # agora, x é uma string
x = [1, 2, 3] # agora, x é uma lista
```

Essa capacidade do Python de mudar a tipagem de uma variável é uma das razões que torna essa linguagem tão rápida na escrita e fácil de ler.

Há uma consequência dessa abordagem de "variável como ponteiro". Se você tiver dois nomes de variáveis apontando para o mesmo objeto mutável, então a mudança em uma mudará a outra. Vejamos um exemplo disso.

```python
# Exemplo 1.3

x = [1, 2, 3]
y = x
print(y)

# Criamos duas variáveis x e y que apontam para o mesmo objeto. 
# Por causa disso, se modificarmos a lista através de um dos nomes, veremos que a "outra" lista também será modificada:

x.append(4) # Adicionamos o valor 4 na lista apontada por x.
print(y) # A lista y foi igualmente alterada!
```

Observe que se usarmos o comando de atribuição = na variável x, isso não afetará o valor de y. Isso ocorre porque o comando = muda apenas o objeto para a qual a variável aponta. Vejamos um exemplo.

```python
# Exemplo 1.4

x = 'alguma coisa' # Agora, x aponta para um objeto diferente (uma string) e não mais para a lista. A lista original permanece inalterada.
print(y) # y permanece inalterada, pois agora x aponta para um objeto diferente (uma string) e não mais para a lista.
```

Você pode se perguntar se esse conceito de ponteiro torna as operações aritméticas em Python difíceis de acompanhar. Mas a linguagem é configurada e tal maneira que essa dificuldade não exista. Valores numéricos, caracteres e outros tipos de dado são imutáveis: você não pode alterar seus valores, apenas os valores para os quais as variáveis apontam. Vejamos outro exemplo. 

```python
# Exemplo 1.5

x = 10
y = x
x += 5 # Adiciona 5 ao valor de x e atribui o resultado a x
print("x =", x)
print("y =", y)
```

Quando digitamos x += 5, não estamos modificando o valor do objeto 5 apontado pela variável x, mas mudando o objeto para o qual x aponta. Por essa razão, o valor de y não é alterado pela operação.

### **<span style="color:green">PRINCÍPIO 2</span>**: Python é uma linguagem de programação orientada a ***objetos***, significando que, em Python, tudo é um objeto.

Vejamos alguns exemplos disso.

```python
# Exemplo 2.1

x = 4
type(x)

# Este é um objeto do tipo INT, ou seja, uma variável quantitativa que representa um número inteiro.
```

```python

x = 'hello'
type(x)

x = 3.14159
type(x)
```
