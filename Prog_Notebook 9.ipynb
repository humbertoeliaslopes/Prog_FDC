### **<span style="color:darkgreen">2) Revisão teórica: o método dos multiplicadores de Lagrange</span>**

O problema de Ana pode ser escrito da seguinte forma:

$$\max_{x,\,y} \; U(x, y) = x^{\alpha} y^{\beta} \quad \text{sujeito a} \quad p_x x + p_y y = M$$

em que $p_x$ e $p_y$ são os preços dos bens e $M$ é a renda. Para resolvê-lo, montamos a função lagrangiana, que incorpora a restrição orçamentária por meio do multiplicador $\lambda$:

$$\mathcal{L}(x, y, \lambda) = x^{\alpha} y^{\beta} + \lambda \, (M - p_x x - p_y y)$$

As condições de primeira ordem exigem que as derivadas parciais de $\mathcal{L}$ sejam iguais a zero:

$$\frac{\partial \mathcal{L}}{\partial x} = \alpha x^{\alpha - 1} y^{\beta} - \lambda p_x = 0$$

$$\frac{\partial \mathcal{L}}{\partial y} = \beta x^{\alpha} y^{\beta - 1} - \lambda p_y = 0$$

$$\frac{\partial \mathcal{L}}{\partial \lambda} = M - p_x x - p_y y = 0$$

Os primeiros termos das duas primeiras condições são as **utilidades marginais** dos bens. Na Cobb-Douglas, elas podem ser escritas de forma simples:

$$UMg_x = \frac{\alpha \, U}{x} \qquad \qquad UMg_y = \frac{\beta \, U}{y}$$

Dividindo a primeira condição pela segunda, obtemos a condição de tangência: no ótimo, a **taxa marginal de substituição** (TMS) é igual à razão entre os preços.

$$TMS = \frac{UMg_x}{UMg_y} = \frac{p_x}{p_y}$$

Substituindo essa condição na restrição orçamentária, chegamos às **funções de demanda** da Cobb-Douglas:

$$x^* = \frac{\alpha}{\alpha + \beta} \times \frac{M}{p_x} \qquad \qquad y^* = \frac{\beta}{\alpha + \beta} \times \frac{M}{p_y}$$

Por fim, a primeira condição nos dá o valor do multiplicador no ótimo:

$$\lambda^* = \frac{UMg_x(x^*, y^*)}{p_x}$$

O multiplicador $\lambda^*$ tem uma interpretação econômica importante: ele mede, aproximadamente, **quanto a utilidade máxima aumenta quando a renda aumenta em R\$ 1,00**. Por isso, é chamado de utilidade marginal da renda.
