### **<span style="color:darkgreen">1) O problema econômico</span>**

Ana acabou de se formar e dispõe de $R\$\;1.200,00$ por mês para gastar com dois bens:

a) **refeições fora de casa** (bem $x$), ao preço de $R\$\;30,00$ cada;

b) **ingressos de cinema e teatro** (bem $y$), ao preço de $R\$\;40,00$ cada.

A satisfação de Ana é medida pela função de utilidade do tipo Cobb-Douglas, a mesma forma funcional que você usou no Exercício 5 do Notebook 3:

$$U(x, y) = x^{\alpha} \times y^{1 - \alpha}$$

com $\alpha = 0{,}6$.

Ana quer comprar a cesta que lhe dá a maior utilidade possível, gastando exatamente a sua renda $M$. Nas aulas de Microeconomia, esse problema é resolvido pelo **método dos multiplicadores de Lagrange**, que leva às seguintes fórmulas:

| O que calcular | Fórmula |
| :---: | :---: |
| Quantidade ótima de refeições | $x^* = \dfrac{\alpha \times M}{p_x}$ |
| Quantidade ótima de ingressos | $y^* = \dfrac{(1 - \alpha) \times M}{p_y}$ |
| Multiplicador de Lagrange | $\lambda^* = \dfrac{U(x^*, y^*)}{M}$ |

Você não precisa deduzir essas fórmulas. A sua tarefa é transformá-las em funções Python. Os valores do problema são hipotéticos.
