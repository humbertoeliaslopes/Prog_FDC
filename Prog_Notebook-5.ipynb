### **<span style="color:darkgreen">2.3) Variáveis String</span>**

As variáveis do tipo string armazenam cadeias de caracteres, ou seja, sequências de letras, números, sinais de pontuação, e assim por diante. O uso mais comum das strings em programação é exibir textos na tela.

Isso faz com que, em Python, precisemos escrever a cadeia de caracteres que desejamos exibir - um texto, por exemplo - entre aspas duplas ou simples. Essa estratégia separa a cadeia dos comandos que devem ser executados em Python. 

Por envolverem uma cadeia de caracteres, as strings têm comprimento, ou tamanho, associado. Sempre é possível verificarmos o comprimento de uma string por meio da função len().

Vejamos alguns exemplos no código a seguir.

```python
# Cria um conjunto de variáveis string e verifica o comprimento de cada uma.

nome = "João"
len(nome)
```

```python
endereco = "Rua Bernardo Guimarães, 3071"
len(endereco)
```

```python
instituicao = "Fundação Dom Cabral (FDC)"
len(instituicao)
```

#### ***<span style="color:darkgreen">2.3.1) Operações com Strings</span>***

> **2.3.1.1) Concatenação**

Os conteúdos de uma variável string podem ser concatenados (somados) por meio do operador "+". Vejamos como isso funciona.

```python
nome = "João"
len(nome)
```

```python
nome = nome + " dos Santos" # Observe que é preciso criar um espaço em branco antes de "dos Santos" para que o nome seja exibido corretamente.
len(nome)
print(nome)
```

```python
nome = nome + " de Oliveira Coutinho"
len(nome)
print(nome)
```

```python
idade = str(23) # Converte o valor inteiro 23 para string e armazena na variável idade.
nome_completo = nome + idade
len(nome_completo)
print(nome_completo)
```

> **2.3.1.2) Composição**

***Composição na Sintaxe Antiga do Python***

Nem sempre é recomendável concatenar várias strings para obter uma mensagem, como o exemplo acima mostra claramente. Justamente por essa razão é que podemos usar o recurso de composição, ativado pelo operador "%". Vamos usá-lo para melhorar o texto da célula anterior.

```python
print("%s tem %d anos de idade." % (nome, idade))

# Observe que os valores das variáveis já estavam armazenados em memória, então não foi necessário digitá-los novamente. A função print() exibiu o conteúdo das variáveis na tela, e a função len() retornou o comprimento de cada uma delas.
```

Os operadores de composição em Python são os seguintes:


| Operador | Tipo |
| :---: | :---: |
| %d | Números inteiros |
| %s | Strings |
| %f | Números decimais |


Vejamos mais exemplos de operações de composição.

```python
nome = "Renata Miranda Teixeira Sobrinho"
print(nome)
len(nome)
```

```python
altura = 1.72
print(altura)
```

```python
print("%s tem %f metros de altura." % (nome, altura))

# Se você rodar esta linha do código, verá que há decimais demais na variável altura. Para limitar o número de casas decimais, você pode usar a seguinte linha de código:
```

```python
print("%s tem %.2f metros de altura." % (nome, altura))

# O algarismo 2 antes do f indica que a variável altura terá apenas duas casas decimais. Se você quiser que a variável altura tenha apenas uma casa decimal, basta substituir o 2 por 1. Se quiser que a variável altura tenha três casas decimais, basta substituir o 2 por 3, e assim por diante.
```

Vamos usar a composição para aprimorar o texto, supondo que temos um conjunto de dados sobre variáveis econômicas.

```python
pib = 12.7
pais = "Brasil"
ano = 2025

"O país %s teve um PIB de %.1f bilhões de dólares em %d." % (pais, pib, ano)
```

***Composição na Nova Sintaxe do Python***

Embora o operador "%" ainda seja válido, as versões mais recentes do Python incorporaram um modo mais simples para compor variáveis string. Basicamente, esse modo substitui o operador pelo comando .format e os %d, %s e %f por {}. 

Vamos reescrever os códigos anteriores usando essa nova sintaxe de composição.

```python
nome = "Renata Miranda Teixeira Sobrinho"
altura = 1.72
"{} tem {:.2f} metros de altura.".format(nome, altura) # O parâmetro .2f ainda é necessário para limitar o número de casas decimais da variável altura.
```

```python
pib = 12.7
pais = "Brasil"
ano = 2025
"O país {} teve um PIB de {:.1f} bilhões de dólares em {}.".format(pais, pib, ano) # O parâmetro .1f ainda é necessário para limitar o número de casas decimais da variável pib.
```

Como podemos observar, a nova sintaxe é mais compacta do que a antiga. Isso significa a possibilidade de termos um código mais parcimonioso, mas executando as mesas operações que desejamos. Isso é melhor para nós.

> **2.3.1.3) Fatiamento**

Em Python, o fatiamento permite dividirmos uma string em pedaços menores. Isso pode facilitar a manipulação desse tipo de dado. Vejamos alguns exemplos.

```python
string_longa = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
print(string_longa)
```

```python
# Agora, vamos localizar as 4 primeiras letras da string_longa. Para isso, você pode usar a seguinte linha de código:

string_longa[0:4] # O índice 0 indica a primeira letra da string_longa, e o índice 4 indica a quinta letra da string_longa. Portanto, a linha de código acima retorna as quatro primeiras letras da string_longa.
```

```python
# Façamos outra operação, agora selecionando as 10 primeiras letras da string_longa. Para isso, você pode usar a seguinte linha de código:  

string_longa[0:10] # O índice 0 indica a primeira letra da string_longa, e o índice 10 indica a décima primeira letra da string_longa. Portanto, a linha de código acima retorna as dez primeiras letras da string_longa.
```

Podemos omitir o primeiro caracter a partir da esquerda, se nossa intenção for listar o que vier após ele:

```python
string_longa[: 7]
```

Também podemos marcar o caracter inicial e omitirmos o final. Python entenderá que pretendemos listar até o último caracter.

```python
string_longa[0: ]
```

```python
Se omitirmos ambos os caracteres, o Python entende que queremos selecionar a string inteira. Portanto, a linha de código acima retorna a string_longa inteira.
```

```python
string_longa[ : ]
```

O fatiamento também pode ser usado com valores negativos, que indicam posições a partir da direita. Em outras palavras, -1 é o último caractere, -2, o penúltimo, e assim por diante.

```python
string_longa[3: -3] # O índice 3 indica a quarta letra da string_longa, e o índice -3 indica a terceira letra da string_longa, contando de trás para frente. Portanto, a linha de código acima retorna a string_longa sem as três primeiras e as três últimas letras.
```

```python
string_longa[-1: 7] # O índice -1 indica a última letra da string_longa, e o índice 7 indica a oitava letra da string_longa. Portanto, a linha de código acima retorna a string_longa sem as sete primeiras letras.
```

**ATENÇÃO**

Os iniciantes em Python frequentemente se perdem nas posições das strings. A primeira posição é sempre 0 (zero), e não 1 (um), como muitos pensam. 

### **<span style="color:darkgreen">2.4) Entrada de Dados</span>**

A entrada de dados permite que um código receba dados ou valores por um dispositivo, como, por exemplo, o teclado do computador. Esse recurso é ativado por meio da função input().

Vejamos alguns exemplos.

```python
valor = input("Digite um valor: ") # A função input() exibe uma mensagem na tela e aguarda que o usuário digite um valor. O valor digitado é armazenado na variável valor.
print("O valor digitado foi: %s" % valor) # A função print() exibe o valor armazenado na variável valor na tela.
```

Vejamos como entrar com dados usando uma das funções mais interessantes em Python: a f string. 

```python
nome = input("Digite seu nome: ") # A função input() exibe uma mensagem na tela e aguarda que o usuário digite um valor. O valor digitado é armazenado na variável nome.
print(f"Seu nome é {nome}") # A função print() exibe o valor armazenado na variável nome e idade na tela.
```

#### ***<span style="color:darkgreen">2.4.1) Conversão da Entrada de Dados</span>***

A função input() sempre retorna valores string, mesmo se digitarmos apenas números. Entretanto, muitas vezes precisamos entrar com valores numéricos de fato, o que demanda a conversão da string gerada pela função input() em números.

Esses números, por sua vez, podem ser inteiros ou com decimais. Assim, para fazer a conversão das strings para o valor numérico correto precisaremos definir se teremos dados do tipo INT ou FLOAT.

Vejamos alguns exemplos.

```python
anos = input("Anos de Trabalho na Empresa") # A função input() exibe uma mensagem na tela e aguarda que o usuário digite um valor. O valor digitado é armazenado na variável anos.
type(anos) # A função type() retorna o tipo de dado da variável anos. No caso, a variável anos é do tipo string (str).
```

Agora, refaçamos o código, transformando a string em valor numérico sem decimais.

```python
anos = int(input("Anos de Trabalho na Empresa"))
print(f"Você tem {anos} anos de trabalho na empresa") # A função print() exibe o valor armazenado na variável anos na tela.
type(anos) # A função type() retorna o tipo de dado da variável anos. No caso, a variável anos é do tipo inteiro (int).
```

Usando a mesma lógica, podemos transformar variáveis string em valores numéricos que aceitam decimais (FLOAT).

```python
altura =input("Digite sua altura em metros: ") # A função input() exibe uma mensagem na tela e aguarda que o usuário digite um valor. O valor digitado é armazenado na variável altura.
type(altura) # A função type() retorna o tipo de dado da variável altura. No caso, a variável altura é do tipo string (str).
```

```python
altura = float(input("Digite sua altura: ")) # A função input() exibe uma mensagem na tela e aguarda que o usuário digite um valor. O valor digitado é armazenado na variável altura.
type(altura) # A função type() retorna o tipo de dado da variável altura. No caso, a variável altura é do tipo float (float).
```
