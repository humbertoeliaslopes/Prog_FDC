### **<span style="color:darkgreen">1) Listas</span>**

Listas são estruturas **mutáveis** que permitem o armazenamento de valores distintos (floats, integers, booleans, etc.), que podem ser acessados por um índice. Elas também podem armazenar zero ou mais elementos de um mesmo tipo ou de tipos diversos, podendo, inclusive, conter outras listas.

Vejamos alguns exemplos de listas.

Suponhamos que temos os seguintes conjuntos de dados:

A = {13, 5, 89, 12, 34} (valores inteiros, ou seja, do tipo INT).

B = { 23,4, 15,8, 33,97} (valores com decimais, ou seja, do tipo FLOAT).

C = (Rio de Janeiro, São Paulo, Belo Horizonte, Porto Alegre) (valores com texto, ou seja, STRING).

D = {TRUE, FALSE, TRUE, TRUE} (valores booleanos)

E = {44,6, Rio de Janeiro, 5, FALSE} (valores misturados, contendo os tipos acima)

Como podemos armazenar esses conjuntos de dados no Python? Supondo que, talvez, queiramos alterar os dados de cada conjunto, devemos escolher uma estrutura de dados mutável, que é a lista. O código a seguir mostra como criar as quatro listas necessárias para os conjuntos de dados desse exemplo.

```python
# Insere os dados nas listas de acordo com o tipo de dado que cada uma deve conter.

conj_a = [13, 5, 89, 12, 34] # Os valores de uma lista devem estar entre colchetes e separados por vírgulas.
conj_b = [23.4, 15.8, 33.97] 
conj_c = ["Rio de Janeiro", "São Paulo", "Belo Horizonte", "Porto Alegre"] # Valores strings devem estar entre aspas simples ou duplas.
conj_d = [True, False, True, True] # Valores booleanos devem ser escritos com a primeira letra maiúscula.
conj_e = [44.6, "Rio de Janeiro", 5, False] # Uma lista pode conter valores de diferentes tipos, como números, strings e booleanos.

# Imprime um relatório com o conteúdo de cada lista, mostrando o tipo de dado que cada uma contém.
print("="*50)
print("Relatório das listas:")
print("="*50)
print("Lista A (inteiros):", conj_a)
print("Lista B (decimais):", conj_b)
print("Lista C (strings):", conj_c)
print("Lista D (booleanos):", conj_d)
print("Lista E (mista):", conj_e)
print("="*50)
```

Como as listas são mutáveis, podemos alterar os dados incluídos nela sempre que quisermos. Mas atenção: para isso usamos o índice da lista, no qual o primeiro elemento recebe sempre o código 0 (zero).

Para entendermos melhor essa explicação, suponhamos que desejamos alterar o primeiro elemento da lista conj_a, que é o valor 13, para 45. 

```python
print(conj_a)
conj_a[0] # Acessa o primeiro elemento da lista conj_a, que é 13.
conj_a[0] = 45 # Altera o valor do primeiro elemento da lista conj_a para 45.
print(conj_a) # Imprime a lista conj_a após a alteração do primeiro elemento.
```

Frequentemente, precisamos conferir o tamanho das listas, para saber se incluímos o número de dados que deveríamos. Para fazer isso, usamos a função len().

```python
len(conj_a) # Retorna o tamanho da lista conj_a, que é 5.
len(conj_b) # Retorna o tamanho da lista conj_b, que é 3.
len(conj_c) # Retorna o tamanho da lista conj_c, que é 4.
len(conj_d) # Retorna o tamanho da lista conj_d, que é 4.
len(conj_e) # Retorna o tamanho da lista conj_e, que é 4.

# Imprime um relatório com o tamanho de cada lista.
print("="*50)
print("Tamanho das listas:")
print("="*50)
print("Lista A (inteiros):", len(conj_a))
print("Lista B (decimais):", len(conj_b))
print("Lista C (strings):", len(conj_c))
print("Lista D (booleanos):", len(conj_d))
print("Lista E (mista):", len(conj_e))
print("="*50)
```

Também podemos adicionar elementos a uma lista por meio da função append().

```python
conj_a.append(67) # Adiciona o valor 67 ao final da lista conj_a.
conj_b.append(12.5) # Adiciona o valor 12.5 ao final da lista conj_b.
conj_c.append("Curitiba") # Adiciona o valor "Curitiba" ao final da lista conj_c.
conj_d.append(False) # Adiciona o valor False ao final da lista conj_d.
conj_e.append("São Paulo") # Adiciona o valor "São Paulo" ao final da lista conj_e.

# Imprime um relatório com o conteúdo de cada lista após a adição dos novos elementos.
print("="*50)
print("Relatório das listas após adição de novos elementos:")
print("="*50)
print("Lista A (inteiros):", conj_a)
print("Lista B (decimais):", conj_b)
print("Lista C (strings):", conj_c)
print("Lista D (booleanos):", conj_d)
print("Lista E (mista):", conj_e)
print("="*50)
```

De maneira oposta, podemos excluir elementos de uma lista por meio da função del() combinada com índices.

```python
del conj_a[2] # Remove o terceiro elemento da lista conj_a, que é 89.
del conj_b[1] # Remove o segundo elemento da lista conj_b, que é 15.8.
del conj_c[0] # Remove o primeiro elemento da lista conj_c, que é "Rio de Janeiro".
del conj_d[3] # Remove o quarto elemento da lista conj_d, que é True.
del conj_e[2] # Remove o terceiro elemento da lista conj_e, que é 5.    

# Imprime um relatório com o conteúdo de cada lista após a remoção dos elementos.
print("="*50)
print("Relatório das listas após remoção de elementos:")
print("="*50)
print("Lista A (inteiros):", conj_a)
print("Lista B (decimais):", conj_b)
print("Lista C (strings):", conj_c)
print("Lista D (booleanos):", conj_d)
print("Lista E (mista):", conj_e)
print("="*50)   
```

### **<span style="color:darkgreen">2) Tuplas</span>**

As tuplas são listas **imutáveis** de dados. Isso significa que, uma vez criadas, não conseguimos alterar os dados incluídos nelas da forma que aprendemos anteriormente. Isso só é possível alterando os valores diretamente no código.

As tuplas são criadas com valores entre parênteses. Vejamos como fazer isso com os valores dos conjuntos que usamos nos exemplos anteriores.

```python
# Insere os dados nas tuplas de acordo com o tipo de dado que cada uma deve conter.

conj_a = (13, 5, 89, 12, 34) # Os valores de uma tupla devem estar entre parênteses e separados por vírgulas.
conj_b = (23.4, 15.8, 33.97) 
conj_c = ("Rio de Janeiro", "São Paulo", "Belo Horizonte", "Porto Alegre") # Valores strings devem estar entre aspas simples ou duplas.
conj_d = (True, False, True, True) # Valores booleanos devem ser escritos com a primeira letra maiúscula.
conj_e = (44.6, "Rio de Janeiro", 5, False) # Uma tupla pode conter valores de diferentes tipos, como números, strings e booleanos.

# Imprime um relatório com o conteúdo de cada tupla, mostrando o tipo de dado que cada uma contém.
print("="*50)
print("Relatório das tuplas:")
print("="*50)
print("Tupla A (inteiros):", conj_a)
print("Tupla B (decimais):", conj_b)
print("Tupla C (strings):", conj_c)
print("Tupla D (booleanos):", conj_d)
print("Tupla E (mista):", conj_e)
print("="*50)
```

Agora, veja o que acontece quando tentamos mudar o valor de uma tupla usando os métodos anteriores.

```python
conj_a(0) = 33 # Tenta alterar o valor do primeiro elemento da tupla conj_a para 33, mas isso não é permitido, pois tuplas são imutáveis.
```

```python
conj_a.append(67) # Adiciona o valor 67 ao final da lista conj_a.
```

```python
del conj_a(1)
```

### **<span style="color:darkgreen">3) Dicionários</span>**

Dicionários são estruturas similares às listas, mas compostos por um conjunto de chaves e valores. Isso facilita muito a digitação de dados que estão relacionados entre si. Vejamos um exemplo.

Desejamos digitar a tabela abaixo. Poderíamos usar listas ou tuplas, mas, uma vez que temos duas colunas com dados diferentes, o dicionário será uma opção mais interessante. 

| Produto | Preço |
| :---: | :---: |
| Alface (unidade) | 1,30 |
| Batata Inglesa (kg) | 4,85 |
| Beterraba (kg) | 3,45 |
| Maçã Fuji (kg) | 11,20 |
| Melão Rei (kg) | 16,90 |

```python
# Cria um dicionário com os dados da tabela acima

tabela = {"Alface (unidade)": 1.30, "Batata Inglesa (kg)": 4.85, "Beterraba (kg)": 3.45, "Maçã Fuji (kg)": 11.20, "Melão Rei (kg)": 16.90} # Os valores de um dicionário devem estar entre chaves e separados por vírgulas. Cada elemento do dicionário é composto por uma chave e um valor, separados por dois pontos.
print(tabela) # Imprime o dicionário tabela.
```

Frequentemente, há muitos dados para digitarmos e, por isso, podemos quebrá-los nas vírgulas:

```python
# Cria um dicionário com os dados da tabela acima

tabela = {"Alface (unidade)": 1.30, 
          "Batata Inglesa (kg)": 4.85,
          "Beterraba (kg)": 3.45, 
          "Maçã Fuji (kg)": 11.20, 
          "Melão Rei (kg)": 16.90} 
print(tabela) # Imprime o dicionário tabela.
```

Podemos acessar os dados de um dicionário usando suas chaves.

```python
print(tabela["Alface (unidade)"]) # Acessa o valor associado à chave "Alface (unidade)" no dicionário tabela, que é 1.30.
```

Se a chave não existir, receberemos uma mensagem de erro.

```python
print(tabela["Manga (kg)"]) # O código acima tenta acessar o valor associado à chave "Manga (kg)" no dicionário tabela, mas essa chave não existe, o que resultará em um erro do tipo KeyError.
```

Dicionários podem ser combinados com listas, ampliando as possibilidades de programação. Vamos entender isso, supondo que um pequeno comerciante queira registrar uma relação de estoque de mercadorias, contendo produto, preço e quantidade em estoque.

```python
estoque = {"Tomate Andreia": [100, 5.60],
           "Cebola": [50, 3.20],
           "Alface": [30, 1.30],
           "Batata Inglesa": [20, 4.85],
           "Beterraba": [15, 3.45],
           "Maçã Fuji": [10, 11.20],
           "Melão Rei": [5, 16.90],}
print(estoque) # Imprime o dicionário estoque.
```

Nesse exemplo, o nome do produto é a chave de localização e a lista consiste nos valores da quantidade disponível e do preço.
